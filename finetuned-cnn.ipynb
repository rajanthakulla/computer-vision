```python
import tensorflow as tf
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
%pip install tensorflow-datasets

#cat vs dog dataset from tfds

import tensorflow_datasets as tfds
(train_ds, val_ds), ds_info = tfds.load('cats_vs_dogs', split=['train[:80%]', 'train[80%:]'], as_supervised=True, with_info=True)
```

```python
IMG_SIZE = 150

def format_img(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))/255.0
    return image, label
```

```python
train_ds = train_ds.map(format_img).batch(32).prefetch(tf.data.AUTOTUNE)
val_ds = val_ds.map(format_img).batch(64).prefetch(tf.data.AUTOTUNE)
```

```python
# Load pretrained VGG16
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(IMG_SIZE, IMG_SIZE, 3))
base_model.trainable = False  # freeze layers
```

```python
# Add custom layers
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history = model.fit(train_ds, validation_data=val_ds, epochs=3)

print("Validation Accuracy:", model.evaluate(val_ds)[1])
```
